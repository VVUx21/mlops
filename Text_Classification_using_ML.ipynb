# Basic Preprocessing
  - Remove tags - HTML
  - Lower case
  - remove stopwords

```python
import re
def remove_tags(raw_text):
    cleaned_text = re.sub(re.compile('<.*?>'), '', raw_text)
    return cleaned_text
```

```python
df['review'] = df['review'].apply(remove_tags)
```

```python
df.head()
```

```python
df['review'][0]
```

```python
df['review'] = df['review'].apply(lambda x:x.lower())
```

```python
df['review'][0]
```

```python
from nltk.corpus import stopwords
import nltk

nltk.download('stopwords')
sw_list = stopwords.words('english')

df['review'] = df['review'].apply(lambda x: [item for item in x.split() if item not in sw_list]).apply(lambda x:" ".join(x))
```

```python
df['review'][0]
```

```python
df.head()
```

```python
X = df.iloc[:,0:1]
y = df['sentiment']
```

```python
X.head()
```

```python
y
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

y = encoder.fit_transform(y)
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=1)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
# Applying BoW
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
```

```python
# Use sparse matrices to save memory - don't convert to dense arrays
X_train_bow = cv.fit_transform(X_train['review'])
X_test_bow = cv.transform(X_test['review'])
```

```python
X_train_bow
```

```python
# Using MultinomialNB instead of GaussianNB - works better with text and sparse matrices
from sklearn.naive_bayes import MultinomialNB
mnb = MultinomialNB()

mnb.fit(X_train_bow, y_train)
```

```python
y_pred = mnb.predict(X_test_bow)

from sklearn.metrics import accuracy_score, confusion_matrix
accuracy_score(y_test, y_pred)
```

```python
confusion_matrix(y_test,y_pred)
```

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier()

rf.fit(X_train_bow,y_train)

y_pred = rf.predict(X_test_bow)

accuracy_score(y_test,y_pred)
```

```python
cv = CountVectorizer(max_features=3000) # Limit features to top 3000 dimensional space

X_train_bow = cv.fit_transform(X_train['review']).toarray()
X_test_bow = cv.transform(X_test['review']).toarray()

rf = RandomForestClassifier()

rf.fit(X_train_bow,y_train)
y_pred = rf.predict(X_test_bow)
accuracy_score(y_test,y_pred)
```

# N grams

```python
cv = CountVectorizer(ngram_range=(1,2),max_features=5000)

X_train_bow = cv.fit_transform(X_train['review']).toarray()
X_test_bow = cv.transform(X_test['review']).toarray()

rf = RandomForestClassifier()

rf.fit(X_train_bow,y_train)
y_pred = rf.predict(X_test_bow)
accuracy_score(y_test,y_pred)
```

TFIDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer()
```
